## **Title**
Markov Decision Process (MDP) Modeling

## **Aim**

To formulate an engineering problem as a Markov Decision Process (MDP) and define its states, actions, rewards, and transitions.

## **Theory**

A Markov Decision Process (MDP) is a mathematical framework used to model decision-making problems in reinforcement learning.

An MDP consists of four main components:

1. States (S):
   Represent the different situations in which the agent can be.

2. Actions (A):
   Represent the choices available to the agent.

3. Rewards (R):
   Represent the feedback received by the agent after taking an action.

4. Transitions (P):
   Represent the probability of moving from one state to another after taking an action.

In this experiment, an engineering problem of a temperature control system is modeled as an MDP.

The objective is to maintain the temperature near the desired value.

The agent observes the current temperature state and chooses an action such as:

- Increase temperature
- Decrease temperature
- Maintain temperature

A positive reward is given when the temperature is close to the desired temperature, while a negative reward is given when the temperature moves away from the desired temperature.

## **Algorithm**

1. Define the states of the temperature control system.
2. Define the actions available to the agent.
3. Define the reward for each state-action combination.
4. Define the transition probabilities between states.
5. Create the MDP representation using a transition table.
6. Display the states, actions, rewards, and transitions.
7. Simulate the system by selecting actions.
8. Display the resulting state transitions and rewards.

## Import Required **Libraries**

```python
import numpy as np
import random
```

## **Define States**

```python
# Define the states of the warehouse robot

states = {
    0: "Charging Station",
    1: "Storage Area",
    2: "Packaging Area",
    3: "Delivery Area"
}

print("States:")
for state, name in states.items():
    print(f"{state}: {name}")
```

## **Define Actions**

```python
# Define possible actions

actions = {
    0: "Move Forward",
    1: "Move Backward",
    2: "Stay"
}

print("Actions:")
for action, name in actions.items():
    print(f"{action}: {name}")
```

## **Define Rewards**

```python
# Define rewards for state-action combinations

rewards = {
    (0, 0): -1,
    (0, 1): -1,
    (0, 2): 2,

    (1, 0): -1,
    (1, 1): -2,
    (1, 2): -1,

    (2, 0): 10,
    (2, 1): -2,
    (2, 2): -1,

    (3, 0): 10,
    (3, 1): -1,
    (3, 2): 0
}

print("Rewards:")
for key, reward in rewards.items():
    state, action = key
    print(
        f"State: {states[state]}, "
        f"Action: {actions[action]}, "
        f"Reward: {reward}"
    )
```

## **Define State Transitions**

```python
# Define deterministic state transitions
# transition[(current_state, action)] = next_state

transition = {
    (0, 0): 1,
    (0, 1): 0,
    (0, 2): 0,

    (1, 0): 2,
    (1, 1): 0,
    (1, 2): 1,

    (2, 0): 3,
    (2, 1): 1,
    (2, 2): 2,

    (3, 0): 3,
    (3, 1): 2,
    (3, 2): 3
}

print("State Transitions:")

for key, next_state in transition.items():
    state, action = key

    print(
        f"{states[state]} + {actions[action]}"
        f" → {states[next_state]}"
    )
```

## **Display MDP Components**

```python
print("========== MDP MODEL ==========")

print("\nStates:")
for state, name in states.items():
    print(f"S{state} = {name}")

print("\nActions:")
for action, name in actions.items():
    print(f"A{action} = {name}")

print("\nRewards:")
for key, reward in rewards.items():
    state, action = key
    print(
        f"R(S{state}, A{action}) = {reward}"
    )

print("\nTransitions:")
for key, next_state in transition.items():
    state, action = key
    print(
        f"P(S{next_state} | S{state}, A{action}) = 1.0"
    )
```

## **Simulate the Robot**

```python
# Simulate the robot moving through the warehouse

current_state = 0
total_reward = 0

print("========== ROBOT SIMULATION ==========")

for step in range(6):

    # Select an action
    action = 0

    # Find next state
    next_state = transition[(current_state, action)]

    # Get reward
    reward = rewards[(current_state, action)]

    print(
        f"Step {step + 1}: "
        f"{states[current_state]} "
        f"--{actions[action]}--> "
        f"{states[next_state]} "
        f"| Reward = {reward}"
    )

    total_reward += reward
    current_state = next_state

print("\nTotal Reward:", total_reward)
```

## **Visualize the MDP States**

```python
import matplotlib.pyplot as plt

state_positions = {
    0: (0, 0),
    1: (2, 0),
    2: (4, 0),
    3: (6, 0)
}

plt.figure(figsize=(12, 3))

for state, (x, y) in state_positions.items():

    plt.scatter(x, y, s=1000)

    plt.text(
        x,
        y,
        f"S{state}",
        ha="center",
        va="center",
        fontsize=12
    )

    plt.text(
        x,
        y - 0.35,
        states[state],
        ha="center",
        fontsize=10
    )

# Draw transitions
for state in range(3):
    x1, y1 = state_positions[state]
    x2, y2 = state_positions[state + 1]

    plt.arrow(
        x1 + 0.25,
        y1,
        x2 - x1 - 0.5,
        0,
        head_width=0.08,
        length_includes_head=True
    )

plt.title("Warehouse Robot MDP")
plt.axis("off")
plt.show()
```
